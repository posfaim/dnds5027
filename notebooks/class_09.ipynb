## Reading files


#### Plain text files

Most the data that we are planning to work with come in files, typically csv files. 

A csv (comma separated values) is a simple text file that looks something like this:
```
Name,Height
Adam,180
Belzebub,220
Papa Smurf,10
...
```
By simple text file we mean that there are no formatting options, no way to insert an image, just plain text. Note that an xls or doc file is not plain text, it is also more complicated to read or write with Python.


Files are typically stored on your computer's hard drive and they are a sequence of 1s and 0s. There are two main types of files:
* Simple text files: when the 1s and 0s translate to characters (examples: csv, txt, html,...)
* Binary files: 1s and 0s could mean anything, including non-textual data (examples: executable file, image, audio, zip,...)

In most cases, working with text files is enough, so here we focus on text files.

To be able to store text as 1s and 0s or recover text from a sequence of 1s and 0s, we need to be able to translate between the two. This is called character encoding ([slides](https://posfaim.github.io/dnds5027/slides/text_encoding.pdf)).

### `with` context manager

Better way to open files: the `with` context manager.

````python
with open('example.csv') as f:
    for line in f:
        print(line, end="")
````

The file is open inside the indented block below the `with` statement. When the program exits the `with` block, it properly closes the file:

````python
f.read()
````

### Parsing the csv example

A csv file is like a table:
* Each row in the file is row in the table
* In each row, the columns are separted by a ','

To work with the data from the file, it has to be parsed. E.g., converted into numbers and stored in variables.

To parse each row, we will use the `.split()` method of strings.

````python
s = "This is a sentence."
words_list=s.split()
words_list
````

````python
row = "apple,12,42,48"
row_list = row.split(',')
row_list
````

Let's store the names and heights of the people in our file in lists:

````python
names = []
heights = []
with open('example.csv') as f:
    f.readline() # the first line in the file is a header, let's skip that
    for line in f:
        l = line.split(',') # split the row into entries
        print(l)
        names.append(l[0]) 
        heights.append(float(l[1])) # we need to convert the height to a number
````

````python
names
````

````python
heights
````

This example already shows that things can get complicated even when parsing csv files. There are python packages that help with this, e.g., `csv` and `pandas`.

### Writing files

We can also write text files:

````python
names
````

````python
with open('output.csv', 'w') as f: #open('example.csv', 'wt') where t stands for text file
    for name in names:
        f.write(name+'\n')
````

The `'w'` passed as argument to the `open()` function means that we are writing, not reading, the file.

When we call the `open('output.csv', 'w')` function, Python creates the `output.csv` file. If the file exists, Python overwrites it (so the previous version is lost!).

You can also append to the end of an existing file:

````python
new_names = ['Humpty Dumpty', 'Ms. Frizzle']
with open('output.csv', 'a') as f:
    for name in new_names:
        f.write(name+'\n')
````

### Character encoding

At the start of this class we saw that a text file is just a sequence of 1s and 0s, and the *character encoding* is the rule that translates between these 1s and 0s and the characters ([slides](https://posfaim.github.io/dnds5027/slides/text_encoding.pdf)).

For plain English text (a–z, digits, punctuation) almost all encodings agree, so we didn't have to worry about this so far. Accented letters (é, ö, ő, ñ, ...) are different: each encoding stores them as different 1s and 0s. If a file is saved with one encoding and opened with another, the accented letters come out garbled, or Python stops with an error.

Today the most widely used encoding is **UTF-8**. It can store every character of every language.

When you don't specify an encoding, `open()` uses your computer's default encoding. You can check what it is:

````python
import locale
locale.getpreferredencoding()
````

* On Mac and Linux, the answer is `'UTF-8'`.
* On Windows, it is often an older encoding, for example `'cp1252'` (Western Europe) or `'cp1250'` (Central Europe), depending on the language settings.

This is why a notebook that works perfectly on a Mac or Linux computer can break on a Windows computer as soon as a file contains an accented letter.

**This is changing in Python 3.15.** Starting with version 3.15, Python uses UTF-8 as the default on every operating system, including Windows. You can check your version with `import sys; sys.version`. Even so, it's good practice to always specify the encoding:
* many computers still have an older Python version installed;
* files made by other people or other programs (e.g. Excel) may use an older encoding.

The solution: **always tell Python the encoding** with the `encoding` argument:

````python
accented_names = ['José', 'Zoë', 'Søren', 'Müller']

with open('accents.csv', 'w', encoding='utf-8') as f:
    for name in accented_names:
        f.write(name + '\n')
````

Reading it back with the same encoding works:

````python
with open('accents.csv', encoding='utf-8') as f:
    print(f.read())
````

#### Simulating Windows

Let's see what happens on a Windows computer (with an older Python version) where the default encoding is `'cp1252'`. There, `open('accents.csv')` without an `encoding` argument behaves exactly like `open('accents.csv', encoding='cp1252')`. So we can imitate a Windows computer on any computer by setting the encoding to `'cp1252'` ourselves:

````python
# simulating Windows: on Windows, open('accents.csv') would do the same
with open('accents.csv', encoding='cp1252') as f:
    print(f.read())
````

Every accented letter became two strange characters (`é` → `Ã©`). If you ever see text like this, the file was most likely saved as UTF-8 but opened with a different encoding.

It can also go the other way. Suppose a classmate on Windows saved a file without specifying the encoding, so it was stored as `'cp1252'`. If we open it as UTF-8, Python stops with a `UnicodeDecodeError`:

````python
# simulating a file saved on Windows: there, open('accents_windows.csv', 'w') would use cp1252
with open('accents_windows.csv', 'w', encoding='cp1252') as f:
    for name in accented_names:
        f.write(name + '\n')

# ... and try to read it as UTF-8
with open('accents_windows.csv', encoding='utf-8') as f:
    print(f.read())
````

**Rules of thumb**
* When uncertain, add `encoding='utf-8'` when you open a text file, both for reading and for writing.
* If you get a `UnicodeDecodeError`, or see characters like `Ã©`, the file was saved in a different encoding. On Windows, try `encoding='cp1252'` (or `'cp1250'` for Central European languages).
* Excel on Windows saves plain "CSV" files in the older encoding. In the *Save as* dialog, choose **CSV UTF-8** instead.
* If you write a csv that someone will open in Excel, use `encoding='utf-8-sig'`. This puts a small invisible marker at the start of the file that tells Excel the file is UTF-8, so the accents display correctly.
